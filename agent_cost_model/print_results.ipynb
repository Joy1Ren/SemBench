```python
# Oracle aggregate metrics table
# Reads from files/{use_case}/metrics/{agent_dir}.json
# Groups by query_id and averages across all runs ({query_id}_{runcount} keys)

from collections import defaultdict

oracle_metrics_path = f"../files/{use_case}/metrics/{agent_dir}.json"

with open(oracle_metrics_path, "r") as f:
    oracle_metrics = json.load(f)

# Group entries by query_id (handles both "13" and "13_1", "13_2", ... keys)
grouped = defaultdict(list)
for key, entry in oracle_metrics.items():
    grouped[str(entry["query_id"])].append(entry)

def safe_avg(entries, field):
    vals = [e[field] for e in entries if field in e and e[field] is not None]
    return sum(vals) / len(vals) if vals else float("nan")

rows = []
for qid in sorted(grouped.keys(), key=lambda x: int(x) if x.isdigit() else x):
    entries = grouped[qid]
    n = len(entries)

    agent_cost   = safe_avg(entries, "agent_cost")
    subset_cost  = safe_avg(entries, "subset_execution_cost")
    oracle_cost  = safe_avg(entries, "oracle_cost")
    total_opt    = agent_cost + subset_cost + oracle_cost

    has_plan_info = any("plans_written" in e for e in entries)

    rows.append({
        "query_id":             qid,
        "n_runs":               n,
        # -- actual plan execution --
        "cost":                 round(safe_avg(entries, "cost"),    6),
        "latency":              round(safe_avg(entries, "latency"), 2),
        "quality":              round(safe_avg(entries, "quality"), 4),
        # -- visual gap --
        " ":                    "",
        # -- optimization info --
        "plans_written":        round(safe_avg(entries, "plans_written"),        2) if has_plan_info else float("nan"),
        "plans_executed":       round(safe_avg(entries, "plans_executed"),       2) if has_plan_info else float("nan"),
        "unique_plans_exec":    round(safe_avg(entries, "unique_plans_executed"),2) if has_plan_info else float("nan"),
        "opt_cost = (agent_cost + subset_exec_cost + oracle_cost)":
            f"{total_opt:.4f}  ({agent_cost:.4f} + {subset_cost:.4f} + {oracle_cost:.4f})",
    })

df_oracle = pd.DataFrame(rows)
pd.set_option("display.max_colwidth", None)
df_oracle
```
